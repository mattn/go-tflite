# go-tflite on gophernotes

Run TensorFlow Lite inference from Go, right in this notebook.

```go
import "github.com/mattn/go-tflite"
```

```go
model := tflite.NewModelFromFile("xor_model.tflite")
options := tflite.NewInterpreterOptions()
interpreter := tflite.NewInterpreter(model, options)
interpreter.AllocateTensors()
```

```go
import "fmt"
for _, v := range [][]float32{{0, 0}, {0, 1}, {1, 0}, {1, 1}} {
    copy(interpreter.GetInputTensor(0).Float32s(), v)
    interpreter.Invoke()
    fmt.Printf("%v xor %v = %v\n", v[0], v[1], int(interpreter.GetOutputTensor(0).Float32s()[0]+0.5))
}
```

The XNNPACK delegate is included as well:

```go
import "github.com/mattn/go-tflite/delegates/xnnpack"
options2 := tflite.NewInterpreterOptions()
options2.AddDelegate(xnnpack.New(xnnpack.DelegateOptions{NumThreads: 2}))
interpreter2 := tflite.NewInterpreter(model, options2)
interpreter2.AllocateTensors()
copy(interpreter2.GetInputTensor(0).Float32s(), []float32{1, 0})
interpreter2.Invoke()
interpreter2.GetOutputTensor(0).Float32s()
```
